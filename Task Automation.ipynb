# Task 3: File Organizer Automation

## CodeAlpha Python Programming Internship

### Objective
To automate the process of organizing files into separate folders based on their file types.

### Features
- Automatically detects file extensions
- Creates folders for different file categories
- Moves files into appropriate folders
- Displays a summary of organized files

### Technologies Used
- Python
- os module
- shutil module
- File Handling

```python
import os
import shutil

print("Libraries imported successfully!")
```

```python
source_folder = "Test files"

print("Source folder:", source_folder)
```

```python
file_categories = {
    "Images": [".jpg", ".jpeg", ".png", ".gif", ".bmp"],
    "Documents": [".pdf", ".doc", ".docx", ".ppt", ".pptx"],
    "Text": [".txt", ".csv"],
    "Audio": [".mp3", ".wav", ".aac"],
    "Videos": [".mp4", ".avi", ".mkv", ".mov"],
    "Archives": [".zip", ".rar", ".7z"]
}

print("File categories defined successfully!")
```

```python
def get_category(file_extension):
    for category, extensions in file_categories.items():
        if file_extension.lower() in extensions:
            return category
    
    return "Others"
```

```python
print(get_category(".jpg"))
print(get_category(".pdf"))
print(get_category(".mp3"))
print(get_category(".xyz"))
```

```python
for category in file_categories:
    folder_path = os.path.join(source_folder, category)
    
    if not os.path.exists(folder_path):
        os.makedirs(folder_path)

# Create folder for unknown file types
others_folder = os.path.join(source_folder, "Others")

if not os.path.exists(others_folder):
    os.makedirs(others_folder)

print("All category folders created successfully!")
```

```python
files_moved = 0

for file_name in os.listdir(source_folder):
    
    file_path = os.path.join(source_folder, file_name)
    
    # Skip folders
    if os.path.isdir(file_path):
        continue
    
    # Skip Jupyter Notebook files
    if file_name.endswith(".ipynb"):
        continue
    
    # Skip the report file
    if file_name == "organization_report.txt":
        continue
    
    # Get file extension
    file_extension = os.path.splitext(file_name)[1]
    
    # Find the correct category
    category = get_category(file_extension)
    
    # Create destination path
    destination_folder = os.path.join(source_folder, category)
    destination_path = os.path.join(destination_folder, file_name)
    
    # Move the file
    shutil.move(file_path, destination_path)
    
    files_moved += 1
    
    print(f"Moved: {file_name} → {category}")

print(f"\nTotal files moved: {files_moved}")
```

```python
print("===== FILE ORGANIZER REPORT =====")

total_files = 0

for category in file_categories:
    folder_path = os.path.join(source_folder, category)
    files = os.listdir(folder_path)
    
    print(f"{category}: {len(files)} file(s)")
    total_files += len(files)

others_folder = os.path.join(source_folder, "Others")
others_files = os.listdir(others_folder)

print(f"Others: {len(others_files)} file(s)")

print("---------------------------------")
print(f"Total organized files: {total_files}")
print("=================================")
```

```python
report_file = os.path.join(source_folder, "organization_report.txt")

with open(report_file, "w") as file:
    file.write("===== FILE ORGANIZER REPORT =====\n\n")
    
    for category in file_categories:
        folder_path = os.path.join(source_folder, category)
        files = os.listdir(folder_path)
        
        file.write(f"{category}: {len(files)} file(s)\n")
    
    file.write(f"Others: {len(others_files)} file(s)\n")
    file.write("---------------------------------\n")
    file.write(f"Total organized files: {total_files}\n")
    file.write("=================================\n")

print("Report saved successfully!")
print("File:", report_file)
```
